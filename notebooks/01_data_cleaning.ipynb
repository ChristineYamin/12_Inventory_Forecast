## NB1 - Data Cleaning Notebook

```python
import pandas as pd
import numpy as np
```

```python
# Load the data
df = pd.read_csv('C:/Projects/12_Inventory_Forecast/data/gold_stock_data.csv')
df.head()
```

```python
# Quick Check
print("Missing values per column:")
print(df.isnull().sum())
```

```python
print("\nStatistical summary (Look for the outlier in 'unit_price'):")
print(df.describe())
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
# Identify the item category for the outlier row
outlier_item = df.loc[df['unit_price'].idxmax(), 'item_name']
median_price = df[df['item_name'] == outlier_item]['unit_price'].median()

# Replace the outlier
df.loc[df['unit_price'] > 5000000, 'unit_price'] = median_price
print(f"Outlier fixed for {outlier_item} using median price: {median_price}")
```

```python
## Impute missing values
df['unit_price'] = df.groupby('item_name')['unit_price'].transform(lambda x: x.fillna(x.median()))

print("Missing values after cleaning:", df.isnull().sum().sum())
```

```python
# Feature engineering
# Add Revenue column
df['revenue'] = df['quantity_sold'] * df['unit_price']

# Add Month column
df['month'] = df['date'].dt.month
```

```python
# Save the cleaned dataset
df.to_csv('cleaned_gold_stock_data.csv', index=False)
print("Cleaning complete. Data saved as cleaned_gold_stock_data.csv")
```
